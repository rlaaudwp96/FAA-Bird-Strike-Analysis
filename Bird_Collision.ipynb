```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


birds = pd.read_csv("birds.csv")
birds.describe()
birds.columns
```

```python
birds_clean = birds.dropna(subset=['phase_of_flt', 'time_of_day', 'state'])
birds_clean
```

```python
birds_CA = birds_clean[birds_clean['state'] == 'CA'].copy()

birds_otherST = birds_clean[birds_clean['state'] != 'CA'].copy()
```

```python
# 1행 2열의 그래프 도화지를 만듭니다. (가로 14, 세로 6 크기)
fig, axes = plt.subplots(1, 2, figsize=(14, 6)) #plt는 matplotlib 라이브러리에서 제공하는 모듈로, 데이터 시각화를 위한 다양한 기능을 제공합니다. subplots() 함수는 하나의 figure 안에 여러 개의 subplot을 생성할 수 있도록 도와줍니다. 이 경우, 1행 2열의 subplot을 생성하여, 두 개의 그래프를 나란히 표시할 수 있도록 합니다. figsize 매개변수는 그래프 도화지의 크기를 설정하며, 가로 14인치, 세로 6인치로 지정되어 있습니다.
#subplot는 matplotlib 라이브러리에서 제공하는 함수로, 하나의 figure 안에 여러 개의 subplot을 생성할 수 있도록 도와줍니다. 이 경우, 1행 2열의 subplot을 생성하여, 두 개의 그래프를 나란히 표시할 수 있도록 합니다..

sns.countplot(data = birds_CA, y = 'phase_of_flt', ax = axes[0], #countplot는 seaborn 라이브러리에서 제공하는 함수로, 범주형 변수의 빈도 수를 시각화하는 막대 그래프를 그리는 데 사용됩니다. 이 경우, 'phase_of_flt' 열의 고유 값과 해당 값의 빈도 수를 계산하여, y축에 해당 값을 표시하고, x축에는 빈도 수를 표시하는 막대 그래프를 그립니다.
              order = birds_CA['phase_of_flt'].value_counts().index, palette = 'viridis') #value counts는 각 범주형 변수의 고유 값과 해당 값의 빈도 수를 계산하는 메서드입니다. 이 경우, 'phase_of_flt' 열의 고유 값과 해당 값의 빈도 수를 계산하여, countplot에서 막대 그래프를 그릴 때 사용됩니다.
axes[0].set_title("Frequency by Phase: CA") #axes[0]는 첫 번째 서브플롯을 나타내며, set_title() 메서드를 사용하여 해당 서브플롯의 제목을 설정합니다. 이 경우, "Frequency by Phase: CA"라는 제목이 첫 번째 서브플롯에 표시됩니다.
#viridis는 색상 팔레트 중 하나로, 데이터 시각화에서 자주 사용되는 색상 맵입니다. 이 팔레트는 색상 변화가 부드럽고, 색맹 사용자에게도 비교적 잘 인식될 수 있도록 설계되었습니다.

sns.countplot(data = birds_otherST, y = 'phase_of_flt', ax = axes[1],
               order = birds_otherST['phase_of_flt'].value_counts().index, palette =  "viridis")
axes[1].set_title("Frequency by Phase: Other States")
#plt는 matplotlib 라이브러리에서 제공하는 모듈로, 데이터 시각화를 위한 다양한 기능을 제공합니다. tight_layout() 함수는 그래프의 레이아웃을 자동으로 조정하여, 서브플롯 간의 간격을 최적화하고, 제목, 축 레이블, 범례 등이 겹치지 않도록 합니다. 이 함수를 호출하면, 그래프가 더 깔끔하게 표시됩니다.
plt.tight_layout() #tight_layout() 함수는 그래프의 레이아웃을 자동으로 조정하여, 서브플롯 간의 간격을 최적화하고, 제목, 축 레이블, 범례 등이 겹치지 않도록 합니다. 이 함수를 호출하면, 그래프가 더 깔끔하게 표시됩니다. 
plt.show() #show() 함수는 현재까지 생성된 모든 그래프를 화면에 표시하는 역할을 합니다. 이 함수를 호출하면, 위에서 설정한 서브플롯과 그래프가 실제로 출력됩니다.

```

```python
fig, axes = plt.subplots(1, 2, figsize = (16, 6))
sns.countplot(data = birds_CA, x = 'phase_of_flt', hue = 'time_of_day', ax = axes[0])
axes[0].set_title("Phase of Flight vs Time of Day (CA)")
axes[0].set_xlabel("Phase Of Flight")
axes[0].set_ylabel("Count")

sns.countplot(data = birds_otherST, x = 'phase_of_flt', hue = 'time_of_day', ax = axes[1])
axes[1].set_title("Phase of Flight vs Time of Day (Other States)")
axes[1].set_xlabel("Phase Of Flight")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
summary_table = birds_clean.groupby(['state', 'phase_of_flt', 'time_of_day']).size().unstack(fill_value=0)

print("\n--- Summary Table: Bird Collisions by State, Phase of Flight, and Time of Day ---")
print(summary_table)
```

```python
phase_order = birds_CA['phase_of_flt'].value_counts().index #이 코드는 birds_CA 데이터프레임에서 'phase_of_flt' 열의 고유 값과 해당 값의 빈도 수를 계산하고, 그 결과를 내림차순으로 정렬한 인덱스를 가져옵니다. 이 인덱스는 이후 그래프를 그릴 때, 'phase_of_flt' 열의 순서를 지정하는 데 사용됩니다.

colors = ['red' if x == 'Approach' else 'lightgray' for x in phase_order] #이 코드는 phase_order 리스트의 각 요소를 순회하면서, 해당 요소가 'Approach'인 경우에는 'red' 색상을, 그렇지 않은 경우에는 'lightgray' 색상을 지정하는 리스트를 생성합니다. 이 리스트는 이후 그래프를 그릴 때, 각 막대의 색상을 지정하는 데 사용됩니다.

sns.countplot(data = birds_CA, 
              y = 'phase_of_flt',
              order = phase_order,
              palette = colors)

#order = 는 countplot에서 막대 그래프를 그릴 때, y축에 표시되는 범주형 변수의 순서를 지정하는 매개변수입니다. 이 경우, phase_order 리스트를 사용하여 'phase_of_flt' 열의 순서를 지정합니다.
#countplot는 seaborn 라이브러리에서 제공하는 함수로, 범주형 변수의 빈도 수를 시각화하는 막대 그래프를 그리는 데 사용됩니다. 이 경우, 'phase_of_flt' 열의 고유 값과 해당 값의 빈도 수를 계산하여, y축에 해당 값을 표시하고, x축에는 빈도 수를 표시하는 막대 그래프를 그립니다.

plt.title("Highlighted CA Bird Collisions: Approach Phase")
plt.show()
```
